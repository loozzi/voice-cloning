```python
import torch
import soundfile as sf
from qwen_tts import Qwen3TTSModel
```

```python
# Load model
model = Qwen3TTSModel.from_pretrained(
    "g-group-ai-lab/gwen-tts-0.6B",
    device_map="cuda:0",
    dtype=torch.float32,
    attn_implementation="eager",
)
```

```python
generation_config = dict(
    temperature=0.3,
    top_k=20,
    top_p=0.9,
    max_new_tokens=480,
    repetition_penalty=2.0,
    subtalker_do_sample=True,
    subtalker_temperature=0.1,
    subtalker_top_k=20,
    subtalker_top_p=1.0,
)
```

```python
text = '''Xin kính chào quý vị, mời quý vị cùng điểm qua những tin tức nổi bật trong ngày.'''
```

```python
wavs, sr = model.generate_voice_clone(
    text=text,
    language="Vietnamese",
    ref_audio="/home/loozzi/voice-cloning/app/voices/an_nhi/audio.wav",
    ref_text="việt nam đang kiêu hãnh bước vào kỷ nguyên vươn mình rực rỡ với khát vọng mãnh liệt, trí tuệ đổi mới và tinh thần đoàn kết đất nước, tự tin bứt phá, kiến tạo một tương lai thịnh vượng và vươn tầm quốc tế.",
    **generation_config,
)

sf.write("output.wav", wavs[0], sr)
```
